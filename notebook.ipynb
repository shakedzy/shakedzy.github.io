```python
import requests
from bs4 import BeautifulSoup
import json

def fetch_medium_posts(username):
    # Step 1: Convert RSS to JSON
    rss_url = f"https://medium.com/feed/@{username}"
    api_url = "https://api.rss2json.com/v1/api.json"
    resp = requests.get(api_url, params={"rss_url": rss_url})
    resp.raise_for_status()
    data = resp.json()

    posts = []
    for item in data.get("items", []):
        title = item.get("title", "")
        content = item.get("content", "")
        snippet = (BeautifulSoup(content, "html.parser")
                   .get_text()[:100] + "…")
        date = item.get("pubDate", "")
        link = item.get("link", "")

        posts.append({
            "title": title,
            "snippet": snippet,
            "date": date,
            "publisher": username,
            "link": link
        })

    # Step 3: Sort descending by date
    posts.sort(key=lambda x: x["date"], reverse=True)
    return posts

# Example usage:
if __name__ == "__main__":
    posts = fetch_medium_posts("shakedzy")
    for p in posts:
        print(p)
```
